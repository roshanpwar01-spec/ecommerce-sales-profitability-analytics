```python
#check the ordeer id 
missing_orders=details.loc[
    ~details['Order ID'].isin(order['Order ID'])
]
print("Missing Order IDs:", len(missing_orders))
missing_orders.head() 
```

```python
#standardize txt column 
order['State']=order['State'].str.title().str.strip()
order['City']=order['City'].str.title().str.strip()

details["Category"] = details["Category"].str.title().str.strip()
details["Sub-Category"] = details["Sub-Category"].str.title().str.strip()
```

```python
#create new feature 
order["Year"] = order["Order Date"].dt.year
```

```python
order["Month"] = order["Order Date"].dt.month_name()
```

```python
order["Month Number"] = order["Order Date"].dt.month
```

```python
order["Day"] = order["Order Date"].dt.day_name()
```

```python
order.head()
```

```python
sales=pd.merge(
    order,details,
    on='Order ID',
    how='inner' 
)
sales.head()
```

```python
sales.shape
```

```python
#profit margin 
sales['Profit Margin (%)']=( 
    sales['Profit']/sales['Amount'])*100
```

```python
sales['Average Price']=(
    sales['Amount']/sales['Quantity']
)
```

```python
order.to_csv("Orders_Cleaned.csv", index=False)
details.to_csv("Order_Details_Cleaned.csv", index=False)
sales.to_csv("Retail_Sales_Cleaned.csv", index=False) 
```
